```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

# -------------------------
# Universe/window U = UA × UB (what complements are taken relative to)
# -------------------------
xmin, xmax, ymin, ymax = -2, 2, -2, 2
Ux0, Ux1 = xmin - 1, xmax + 1
Uy0, Uy1 = ymin - 1, ymax + 1

# A = [0,1], B = [0,1]
A0, A1 = 0, 1
B0, B1 = 0, 1

# -------------------------
# Helpers
# -------------------------
def style_axes(ax):
    ax.set(xlim=(Ux0, Ux1), ylim=(Uy0, Uy1), aspect="equal")
    ax.spines["bottom"].set_position("zero")
    ax.spines["left"].set_position("zero")
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

    ax.set_xlabel("A", size=20, labelpad=-22, x=0.98)
    ax.set_ylabel("B", size=20, labelpad=-34, y=0.95, rotation=0)

    x_ticks = np.arange(xmin, xmax + 1, 1)
    y_ticks = np.arange(ymin, ymax + 1, 1)
    ax.set_xticks(x_ticks[x_ticks != 0])
    ax.set_yticks(y_ticks[y_ticks != 0])

    ax.grid(True, alpha=0.15)

def draw_AxB(ax):
    # Filled red square for A×B
    ax.add_patch(Rectangle(
        (A0, B0), A1 - A0, B1 - B0,
        facecolor="red", alpha=0.25,
        edgecolor="black", linewidth=2
    ))
    ax.text(0.5, 0.5, r"$A\times B$", ha="center", va="center", fontsize=20)

def shade_AxB_complement(ax, facecolor="grey", alpha=0.18):
    # (A×B)^c = U \ (A×B) = 4 rectangles (left, right, bottom, top)
    patches = [
        Rectangle((Ux0, Uy0), A0 - Ux0, Uy1 - Uy0),          # left
        Rectangle((A1,  Uy0), Ux1 - A1, Uy1 - Uy0),          # right
        Rectangle((A0,  Uy0), A1 - A0, B0 - Uy0),            # bottom middle
        Rectangle((A0,  B1 ), A1 - A0, Uy1 - B1),            # top middle
    ]
    for p in patches:
        p.set(facecolor=facecolor, alpha=alpha, edgecolor="none")
        ax.add_patch(p)

def shade_Ac_x_Bc(ax, facecolor="grey", alpha=0.35):
    # A^c×B^c = (UA\A)×(UB\B) = four corner rectangles
    patches = [
        Rectangle((Ux0, Uy0), A0 - Ux0, B0 - Uy0),           # left-bot
        Rectangle((Ux0, B1 ), A0 - Ux0, Uy1 - B1),           # left-top
        Rectangle((A1,  Uy0), Ux1 - A1, B0 - Uy0),           # right-bot
        Rectangle((A1,  B1 ), Ux1 - A1, Uy1 - B1),           # right-top
    ]
    for p in patches:
        p.set(facecolor=facecolor, alpha=alpha, edgecolor="none")
        ax.add_patch(p)

# -------------------------
# Plot
# -------------------------
fig, ax = plt.subplots(1, 2, figsize=(10, 5), constrained_layout=True)

for a in ax:
    style_axes(a)
    draw_AxB(a)

# Left panel: (A×B)^c
shade_AxB_complement(ax[0], alpha=0.18)
ax[0].set_title(r"$(A\times B)^c$", fontsize=20)

# Right panel: A^c × B^c
shade_Ac_x_Bc(ax[1], alpha=0.35)
ax[1].set_title(r"$A^c\times B^c$", fontsize=20)

# Optional: label the “missing” cross-arms on the left to emphasize ≠
ax[0].text(0.5, Uy0 + 0.4, r"$A\times B^c$", ha="center", va="bottom", fontsize=16)
ax[0].text(Ux0 + 0.5, 0.5, r"$A^c\times B$", ha="left", va="center", fontsize=16, rotation=90)

plt.show()
```
